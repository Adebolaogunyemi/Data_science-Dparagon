#### Problem Statement

You will train your model on 70% of the data and test your model on the final 30% of the data, across four East African countries - Kenya, Rwanda, Tanzania, and Uganda.

Data Understanding.

The data provided presents the countries , the year, each country unique id , their locations and corresponding bank account.

```python
x = df.drop('bank_account', axis=1)
y = df.bank_account
```

```python
x
```

### Encoding
- Label Encoding
- One hot Encoding

replace does something similar to label encoding

```python
#### Feature Selection

X_selected = x.drop(['country', 'year', 'uniqueid', 'gender_of_respondent'], axis=1)
```

```python
X_selected
```

```python
from sklearn.preprocessing import LabelEncoder

labelEncoder = LabelEncoder()
```

```python
X_selected['location_type'] = labelEncoder.fit_transform(X_selected['location_type'])
X_selected['cellphone_access'] = labelEncoder.fit_transform(X_selected['cellphone_access'])
X_selected['relationship_with_head'] = labelEncoder.fit_transform(X_selected['relationship_with_head'])
X_selected['marital_status'] = labelEncoder.fit_transform(X_selected['marital_status'])
X_selected['education_level'] = labelEncoder.fit_transform(X_selected['education_level'])
X_selected['job_type'] = labelEncoder.fit_transform(X_selected['job_type'])
```

```python
X_selected
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X_selected, y, test_size=0.3, random_state=0, stratify=y)
```

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression()
```

```python
model.fit(X_train, y_train)
```

```python
from sklearn.metrics import accuracy_score, classification_report

predictions = model.predict(X_test)
```

```python
print(classification_report(y_test, predictions))
```

```python
print(f'Accuracy score : {accuracy_score(y_test,predictions)}')
```
